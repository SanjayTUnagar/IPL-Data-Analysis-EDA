# 12. Correlation Analysis

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Select numeric columns
numeric_df = df[["year", "innings"]]

# Calculate correlation
print(numeric_df.corr())

# Visualize correlation matrix
sns.heatmap(
    numeric_df.corr(),
    annot=True,      # Show correlation values
    cmap="coolwarm", # Color map
    fmt=".2f"        # Display 2 decimal places
)

plt.title("Correlation Heatmap")
plt.show()
```

# 14. Feature Engineering for IPL Dataset

## 1. Extract Match Year

### Question: Which year was the match played?

```python
df["match_year"] = pd.to_datetime(df["date"]).dt.year

df[["date", "match_year"]].head()
```

## 2. Extract Match Month

### Question: In which month was the match played?

```python
df["match_month"] = pd.to_datetime(df["date"]).dt.month

df[["date", "match_month"]].head()

```

### Question: On which day of the week are most IPL matches played?

```python
df["day_name"].value_counts()

```

## Visualization:

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(data=df, x="day_name",
              order=df["day_name"].value_counts().index)

plt.title("IPL Matches by Day of Week")
plt.xticks(rotation=45)
plt.show()
```
